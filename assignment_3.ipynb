Project Problem

Machine learning based prediction of roof fall in underground mines: Morupule coal Mine case study.

Roof falls are a major geotechnical hazard that threaten work, and production safety. Roof stability is influenced be several geological, geotechnical,and mining factors, making it difficult to accurately predict when and where roof failure may occur using conventional empirical approaches alone limited by them being formulated with a lot of assumtions.

Intended ML outcome

To develop a machine learning model that can use relevent geological, geotechnical and operational parameters to predict the likelihood of roof fall risk such as stable, warning, or high risk conditions.









Data Required

Geological Data- rock type,faults etc
Geotechnical Data- rock strength,rock mass quality etc
Mining and Operational Data- pillar dimensions, depth to cover etc
Ground support- bolt spacing, bolt length, type of support etc
Roof History Data- records of previous roof fall events
